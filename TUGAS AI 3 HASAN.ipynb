## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
`\frac{\partial T}{\partial t}`{=tex}=`\alpha`{=tex}`\frac{\partial^2T}{\partial x^2}`{=tex}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (`\alpha`{=tex})
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=`\frac{\alpha\Delta t}{(\Delta x)^2}`{=tex} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r`\leq 1`{=tex}/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

## 4. Recipe Analisis Spektrum Frekuensi Sinyal Seismik Menggunakan FFT

- **Recipe ID:** `R-SIG-001`
- **Prompt ID:** `P-SIG-001`
- **Status evidence:** `Draft / Not yet executed`
- **Tujuan utama:** mengolah sinyal seismik dalam domain waktu dan frekuensi untuk mengenali komponen frekuensi dominan serta menghasilkan grafik spektrum yang dapat diperiksa.
- **Related artifact:** `code/recipes/signal/r-sig-001.py` (path rancangan)

### Problem

Data seismik umumnya direkam sebagai amplitudo terhadap waktu. Melihat bentuk gelombang saja belum tentu cukup untuk mengetahui komponen frekuensi yang menyusun sinyal. Recipe ini merancang alur kerja untuk membaca data sinyal, memeriksa kualitasnya, menghitung Fast Fourier Transform (FFT), lalu menampilkan spektrum amplitudo terhadap frekuensi.

Analisis ini dapat digunakan sebagai pemeriksaan awal karakteristik sinyal, tetapi hasil spektrum tidak boleh langsung dianggap sebagai bukti jenis batuan atau sumber sinyal tertentu tanpa informasi pendukung.

### Minimal Theory

FFT merupakan algoritma untuk menghitung Discrete Fourier Transform (DFT) secara efisien. DFT mengubah deret sampel dari domain waktu menjadi representasi domain frekuensi.

Untuk sinyal diskret \(x[n]\) dengan jumlah sampel \(N\), DFT dapat ditulis sebagai:

\[X[k]=\sum_{n=0}^{N-1}x[n]e^{-i2\pi kn/N}\]

Jika interval sampling adalah \(\Delta t\), frekuensi samplingnya:

\[f_s=rac{1}{\Delta t}\]

Jarak antarbin frekuensi untuk data dengan panjang \(N\) adalah \(\Delta f=f_s/N\). Untuk sinyal real, spektrum satu sisi umumnya cukup ditampilkan dari 0 hingga frekuensi Nyquist, yaitu \(f_s/2\).

Sebelum FFT, data perlu diperiksa agar interval sampling konsisten dan nilai hilang atau tidak valid diketahui. Pengurangan nilai rata-rata dapat membantu mengurangi komponen DC. Window, misalnya Hann, dapat digunakan untuk mengurangi kebocoran spektral; pilihan ini harus dicatat karena memengaruhi amplitudo spektrum. Frekuensi dominan adalah puncak spektrum pada rentang yang dianalisis, bukan otomatis frekuensi sumber geologi.

### Input dan Constraints

Input utama:
- `DATA_PATH`: lokasi file data sinyal.
- `TIME_COLUMN`: nama kolom waktu, jika tersedia.
- `AMPLITUDE_COLUMN`: nama kolom amplitudo.
- `SAMPLING_RATE_HZ`: frekuensi sampling dalam Hz, jika tidak dihitung dari kolom waktu.
- `WINDOW_TYPE`: jenis window, misalnya `hann` atau `none`.

Batasan:
1. File mentah dipertahankan dan dibaca secara *read-only*.
2. Data harus memiliki amplitudo numerik dan jumlah sampel yang memadai.
3. Jika waktu tersedia, interval sampling harus diperiksa konsistensinya.
4. Jangan menebak frekuensi sampling ketika informasi waktu maupun sampling rate tidak tersedia.
5. Nilai hilang dan nilai nonnumerik harus dilaporkan; jangan dihapus diam-diam.
6. Spektrum harus mencantumkan satuan frekuensi dan metode pemrosesan.
7. Hasil FFT tidak cukup untuk menentukan litologi atau mekanisme sumber secara mandiri.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda berperan sebagai asisten pengolahan sinyal geofisika. Baca `[DATA_PATH]` secara read-only dan gunakan kolom waktu `[TIME_COLUMN]` serta amplitudo `[AMPLITUDE_COLUMN]` jika tersedia. Periksa struktur data, jumlah sampel, nilai hilang, nilai nonnumerik, serta konsistensi interval sampling. Jika frekuensi sampling diberikan, gunakan `[SAMPLING_RATE_HZ]` Hz dan nyatakan sumber nilai tersebut.
>
> Setelah pemeriksaan kualitas data, kurangi nilai rata-rata bila sesuai, terapkan window `[WINDOW_TYPE]` sesuai konfigurasi, lalu hitung FFT. Buat grafik sinyal pada domain waktu dan spektrum amplitudo satu sisi pada domain frekuensi. Laporkan frekuensi sampling, jumlah sampel, resolusi frekuensi, rentang frekuensi yang ditampilkan, serta frekuensi puncak yang teridentifikasi. Jangan mengklaim interpretasi geologi yang tidak didukung oleh data.
>
> Simpan tabel spektrum, grafik, dan laporan QA/QC pada path output yang ditentukan. Catat konfigurasi dan status eksekusi. Jika program belum benar-benar dijalankan, tulis `Not yet executed` dan jangan menyatakan hasil sebagai hasil pengukuran yang sudah terverifikasi.

**Procedure:**
1. Baca file input dan catat nama file serta konfigurasi.
2. Periksa kolom, tipe data, jumlah sampel, nilai hilang, dan nilai nonnumerik.
