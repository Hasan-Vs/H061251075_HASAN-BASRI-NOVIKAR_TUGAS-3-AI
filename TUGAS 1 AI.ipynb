# Prompt 1 — Identifikasi Zona Sesar Menggunakan Metode Gravitasi

### R — Role
Anda berperan sebagai **ahli geofisika struktur dan interpretasi data gravitasi**.

### T — Task
Analisis anomali gravitasi untuk mengidentifikasi kelurusan dan batas kontras densitas yang mungkin berkaitan dengan zona sesar.

### C — Context
Suatu daerah memperlihatkan kelurusan morfologi dan indikasi deformasi geologi. Survei gravitasi dilakukan untuk mempelajari perubahan lateral densitas batuan bawah permukaan.

### I — Input
Data yang diberikan dapat berupa:
- Koordinat dan elevasi stasiun.
- Nilai anomali Bouguer lengkap.
- Peta anomali regional dan residual jika tersedia.
- Peta geologi dan informasi sesar permukaan.

### K — Constraints
- Jangan menyamakan setiap gradien gravitasi dengan sesar.
- Jangan mengarang densitas atau kedalaman.
- Pertimbangkan efek topografi dan alternatif kontak litologi.

### O — Output
Berikan:
1. Peta atau deskripsi pola anomali.
2. Identifikasi gradien dan kelurusan.
3. Interpretasi kemungkinan zona sesar.
4. Tabel bukti dan alternatif interpretasi.
5. Rekomendasi lintasan survei lanjutan.

### Q — Quality Criteria
Interpretasi konsisten dengan prinsip gravitasi, geologi setempat, dan resolusi data.

### V — Verification
Periksa koreksi gravitasi, satuan mGal, koordinat, arah gradien, dan kecocokan dengan peta geologi.

# Prompt 2 — Eksplorasi Potensi Panas Bumi

### R — Role
Anda berperan sebagai **ahli geofisika eksplorasi panas bumi**.

### T — Task
Integrasikan data magnetik dan resistivitas untuk mengenali indikasi struktur pengontrol dan zona alterasi hidrotermal.

### C — Context
Di sekitar manifestasi mata air panas, survei geofisika dilakukan untuk menyusun model konseptual awal sistem panas bumi.

### I — Input
Data yang diberikan dapat berupa:
- Peta anomali magnetik.
- Penampang resistivitas atau magnetotelurik jika tersedia.
- Koordinat manifestasi panas bumi.
- Peta geologi dan struktur.
- Data temperatur atau geokimia jika tersedia.

### K — Constraints
- Resistivitas rendah tidak otomatis menunjukkan reservoir.
- Jangan menentukan suhu atau kapasitas energi tanpa data.
- Pertimbangkan pengaruh lempung, litologi, dan air tanah.

### O — Output
Berikan:
1. Deskripsi anomali magnetik dan resistivitas.
2. Interpretasi kemungkinan alterasi dan struktur.
3. Model konseptual awal.
4. Tabel zona dan tingkat ketidakpastian.
5. Usulan pengukuran lanjutan.

### Q — Quality Criteria
Penafsiran menggabungkan lebih dari satu bukti dan memisahkan pengamatan dari hipotesis.

### V — Verification
Cocokkan anomali dengan struktur geologi, manifestasi, kedalaman investigasi, dan kemungkinan penyebab nontermal.

# Prompt 3 — Analisis Struktur Bawah Permukaan Gunung Api

### R — Role
Anda berperan sebagai **ahli seismologi gunung api**.

### T — Task
Analisis distribusi hiposentrum dan waktu tiba gelombang seismik untuk mempelajari pola aktivitas dan kemungkinan struktur bawah permukaan gunung api.

### C — Context
Jaringan seismometer merekam gempa vulkanik di sekitar gunung api aktif. Data digunakan untuk memetakan sebaran sumber gempa dan mengenali pola spasialnya.

### I — Input
Data yang diberikan dapat berupa:
- Waktu tiba gelombang P dan S.
- Koordinat stasiun.
- Katalog hiposentrum dan magnitudo jika tersedia.
- Model kecepatan yang digunakan.
- Informasi geologi gunung api.

### K — Constraints
- Jangan menyatakan lokasi magma secara pasti dari sebaran gempa saja.
- Jangan memprediksi waktu erupsi.
- Jelaskan ketidakpastian lokasi dan keterbatasan jaringan stasiun.

### O — Output
Berikan:
1. Peta sebaran episentrum dan penampang kedalaman.
2. Pola spasial dan temporal gempa.
3. Interpretasi struktur yang mungkin berkaitan.
4. Tabel ringkasan kejadian.
5. Kesimpulan beserta keterbatasan.

### Q — Quality Criteria
Analisis memakai parameter seismologi yang benar dan tidak melampaui dukungan data.

### V — Verification
Periksa konsistensi waktu tiba, residual lokasi, model kecepatan, kedalaman, dan cakupan azimut stasiun.

# Prompt 4 — Identifikasi Kerentanan Likuefaksi Menggunakan Mikrotremor

### R — Role
Anda berperan sebagai **ahli geofisika teknik dan mikrozonasi seismik**.

### T — Task
Analisis pengukuran mikrotremor untuk mengidentifikasi frekuensi dominan dan indikasi amplifikasi tapak sebagai masukan awal kajian kerentanan likuefaksi.

### C — Context
Kawasan dataran aluvial akan dikembangkan untuk permukiman. Pengukuran mikrotremor dilakukan untuk memahami respons dinamis tanah, kemudian dibandingkan dengan data geoteknik.

### I — Input
Data yang diberikan dapat berupa:
- Rekaman mikrotremor tiga komponen.
- Kurva HVSR dan frekuensi dominan.
- Data geologi permukaan.
- Data muka air tanah, SPT atau CPT jika tersedia.
- Koordinat pengukuran.

### K — Constraints
- HVSR saja tidak dapat membuktikan potensi likuefaksi.
- Jangan mengubah puncak HVSR langsung menjadi faktor keamanan likuefaksi.
- Pisahkan respons tapak dari sifat tanah jenuh dan pemicu gempa.

### O — Output
Berikan:
1. Peta atau tabel frekuensi dominan.
2. Identifikasi lokasi yang memerlukan investigasi lanjut.
3. Perbandingan dengan geologi dan data geoteknik.
4. Pembahasan keterbatasan HVSR.
5. Rekomendasi uji tambahan.

### Q — Quality Criteria
Kesimpulan membedakan indikasi respons tapak dan penilaian likuefaksi yang membutuhkan data geoteknik.

### V — Verification
Periksa kualitas rekaman, kestabilan puncak HVSR, kesesuaian geologi, dan ketersediaan parameter SPT/CPT serta muka air tanah.

# Prompt 5 — Pemetaan Ketebalan Sedimen Menggunakan Seismik Refraksi

### R — Role
Anda berperan sebagai **ahli geofisika dekat permukaan dan seismik refraksi**.

### T — Task
Interpretasikan kurva waktu tempuh untuk memperkirakan kecepatan lapisan dan kedalaman batas sedimen–batuan dasar.

### C — Context
Sebuah lembah sedimen akan dikaji untuk kebutuhan perencanaan tata ruang. Survei seismik refraksi dilakukan sepanjang beberapa lintasan untuk memperkirakan variasi ketebalan sedimen.

### I — Input
Data yang diberikan dapat berupa:
- Jarak sumber–geofon.
- Waktu tiba pertama gelombang P.
- Geometri lintasan dan elevasi.
- Kurva waktu tempuh.
- Informasi singkapan atau bor jika tersedia.

### K — Constraints
- Jangan mengasumsikan semua lapisan makin cepat tanpa menguji keterbatasan metode.
- Jangan mengarang kedalaman lapisan.
- Jelaskan kemungkinan lapisan tersembunyi dan ketidakunikan inversi.

### O — Output
Berikan:
1. Grafik waktu tempuh.
2. Perkiraan kecepatan tiap lapisan.
3. Penampang ketebalan sedimen.
4. Tabel parameter dan ketidakpastian.
5. Kesimpulan dan kebutuhan validasi.

### Q — Quality Criteria
Perhitungan menggunakan satuan konsisten dan penafsiran sesuai geometri akuisisi.

### V — Verification
Periksa picking first break, jarak dan elevasi, kecocokan kurva model terhadap observasi, serta validasi data bor bila tersedia.
